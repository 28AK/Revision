### 1. Find the smallest number in an array

```python
def smallest(arr):
    minimum = arr[0]
    for i in range(len(arr)):
        if arr[i] < minimum:
            minimum = arr[i]
    return minimum

n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)

print(smallest(arr))
```

### 2. Find the largest number in an array

```python
def largest(arr):
    maximum = arr[0]
    for i in range(len(arr)):
        if arr[i] > maximum:
            maximum = arr[i]
    return maximum

n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)

print(largest(arr))
```

### 3. Third maximum number in an array

```python
def third_max(arr):
    arr = list(set(arr))
    arr.sort()
    if len(arr) >= 3:
        return arr[-3]
    else:
        return arr[-1]

n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)

print(third_max(arr))
```

### 4. Second smallest and second largest element in an array

```python
def second_smallest_largest(arr):
    smallest = float('inf')
    second_smallest = float('inf')
    largest = float('-inf')
    second_largest = float('-inf')
    for num in arr:
        if num < smallest:
            second_smallest = smallest
            smallest = num
        elif num < second_smallest and num != smallest:
            second_smallest = num
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest and num != largest:
            second_largest = num

    return second_smallest , second_largest

    
n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)

print(second_smallest_largest(arr))
```

### 5. Reverse a given array(Brute Force)

```python
def reverse(arr):
    result = []
    for i in range(len(arr)-1, -1 , -1):
        result.append(arr[i])
    return result


n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)
    
print(reverse(arr))
```

### 5. Reverse a given array(Two Pointers Approach)

```python
def reverse(arr):
    left = 0
    right = len(arr)-1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr

n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)
    
print(reverse(arr))
```

### 6. Count frequency of each element in an array(Brute Force)

```python
def count_freq(arr):
    for i in range(len(arr)):
        count = 0
        for j in range(len(arr)):
            if arr[i] == arr[j]:
                count += 1
        print(arr[i] ,"->", count)

n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)

print(count_freq(arr))
```

### 6. Count frequency of each element in an array(Hashing)

```python
def count_freq(arr):
    freq = {}
    for num in arr:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
    return freq

n = int(input("Enter size of arr: "))
arr = []
for i in range(n):
    num = int(input("Enter num: "))
    arr.append(num)

print(count_freq(arr))
```

### 6. Unique Number of Occurrences(Leetcode ques)

```python
class Solution:
    def uniqueOccurrences(self, arr: list[int]) -> bool:
        freq = {}
        for num in arr:
            if num in freq:
                freq[num] += 1
            else:
                freq[num] = 1
        frequencies = freq.values()
        return len(frequencies) == len(set(frequencies))
```
